## Indexing: Text Embedding with OpenAI

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters.markdown import MarkdownHeaderTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_openai.embeddings import OpenAIEmbeddings
import numpy as np
```

```python
loader_pdff = PyPDFLoader("Downloads/Introduction_to_Data_and_Data_Science_2.pdf")
pagee_pdf = loader_pdf.load()

for i in range(len(pagee_pdf)):
    pagee_pdf[i].page_content = " ".join(pagee_pdf[i].page_content.split())

characterr_spltter = CharacterTextSplitter(separator = ".", chunk_size = 500, 
                                          chunk_overlap = 50)
pagess_character_splitter = characterr_spltter.split_documents(pagee_pdf)

```

```python
#Embedding/ Intance of open ai class
embedding = OpenAIEmbeddings(model = "text-embedding-ada-002",
                            api_key=os.getenv("OPEENAI_API_KEY"))
```

```python
vector1 = embedding.embed_query(pagess_character_splitter[3].page_content)
vector2 = embedding.embed_query(pagess_character_splitter[5].page_content)
vector3 = embedding.embed_query(pagess_character_splitter[18].page_content)
```

```python
pagess_character_splitter[5].page_content
```

```python
len(vector1), len(vector2), len(vector3)
```

```python
#applying dot-product to each pair of 
#vector and quantifying their similarity using Numpy
np.dot(vector1, vector2),  np.dot(vector1, vector3), np.dot(vector2, vector3)
```

```python
#making sure the vector created by
#openai embedding has a magnitude of 1 
#Using Numpy Linear Algbra
np.linalg.norm(vector1), np.linalg.norm(vector2), np.linalg.norm(vector3)
```

```python
%load_ext dotenv
%dotenv
```

```python
#Constructing retriever chain
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.prompts import PromptTemplate
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.runnables.passthrough import RunnablePassthrough
from langchain_core.runnables import RunnableParallel
from langchain_core.output_parsers import StrOutputParser
```

```python
#embedding
embedding = OpenAIEmbeddings(model = "text-embedding-ada-002",
                            api_key=os.getenv("OPEENAI_API_KEY"))

vectorstore_from_directory = Chroma( persist_directory = "Downloads/Intro_to_ds_lecture",
                                     embedding_function = embedding,)
```

```python
len(vectorstore.get()["documents"])
```

```python
#Retriever
retriever = vectorstore.as_retriever(search_type = "mmr",
                                    search_kwargs = {"k": 3, 
                                                     "lambda_mult":0.7})
```

```python
#prompt template
TEMPLATE = """
Answer the following question:
{question}

Use only the information provided in the following context to answer the question:
{context}

If the context does not contain enough information to answer the question,
clearly state that you do not have enough information. Do not make up facts.
"""
prompt_template = PromptTemplate.from_template(TEMPLATE)
```

```python
#chat open ai
chaat = ChatOpenAI(model_name = "gpt-4.1",
                  api_key=os.getenv("OPEENAI_API_KEY"),
                  seed = 365,
                  temperature= 0,
                  max_tokens= 250)
```

```python
#input
Questions = "What Software, programming language and technique do Data scientist use?"
```

```python
chain = ({"context": retriever,
        "question": RunnablePassthrough()} 
         | prompt_template 
         |  chaat )
```

```python
chain.invoke(Questions)
```
